```python
import matplotlib.pyplot as plt
import numpy as np

# Data
# Riemannian
rzz_riemann = [27, 39]
loss_riemann = [8.311258906190666e-05, 6.137574337650165e-07]

# Ising HW Friendly
rzz_ising = [26, 38]
loss_ising = [0.0006583791666388006, 0.0001261573243568348]

# Trotter (same x-axis as ising hw friendly)
rzz_trotter = [26, 38]
loss_trotter = [0.0027119098576176004, 0.0005299578732859134]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(rzz_riemann, loss_riemann, 'o-', color='royalblue', linewidth=2,
        markersize=8, label='Riemannian')
ax.plot(rzz_ising, loss_ising, 's-', color='darkorange', linewidth=2,
        markersize=8, label='Ising HW Friendly')
ax.plot(rzz_trotter, loss_trotter, '^--', color='forestgreen', linewidth=2,
        markersize=8, label='Trotter')

ax.set_yscale('log')
ax.set_xlabel('Number of RZZ Layers', fontsize=13)
ax.set_ylabel('Loss', fontsize=13)
ax.set_title('Loss vs. Number of RZZ Layers', fontsize=14)
ax.legend(fontsize=11)
ax.grid(True, which='both', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
