# Rotations of Single Qubits

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def rotate_vector(vector, unit_axis, theta):
    # Rodrigues rotation formula
    cos_theta = np.cos(theta)
    sin_theta = np.sin(theta)
    
    # Calculate rotation matrix components
    term1 = cos_theta * vector
    term2 = sin_theta * np.cross(unit_axis, vector)
    term3 = (1 - cos_theta) * np.dot(unit_axis, vector) * unit_axis
    
    return term1 + term2 + term3

# Create sample data
vector = np.array([1, 0, 0])  # Vector to rotate
unit_axis = np.array([0, 0, 1])  # Rotation axis (z-axis)
theta = np.pi/4  # 45 degrees rotation

# Perform rotation
rotated_vector = rotate_vector(vector, unit_axis, theta)

# Plotting
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot original and rotated vectors
ax.quiver(0, 0, 0, vector[0], vector[1], vector[2], color='r', label='Original')
ax.quiver(0, 0, 0, rotated_vector[0], rotated_vector[1], rotated_vector[2], color='b', label='Rotated')
ax.quiver(0, 0, 0, unit_axis[0], unit_axis[1], unit_axis[2], color='g', label='Rotation Axis')

# Set plot properties
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Vector Rotation')
ax.legend()
ax.set_xlim([-1.5, 1.5])
ax.set_ylim([-1.5, 1.5])
ax.set_zlim([-1.5, 1.5])

plt.show()
```
